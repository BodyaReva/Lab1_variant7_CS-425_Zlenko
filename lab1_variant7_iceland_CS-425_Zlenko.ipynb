# Завдання 4. Переведення числа з десяткової системи числення у двійкову

Функція task_7() зчитує число (ціле або з рухомою комою; роздільник — крапка або кома) та виводить його у двійковій системі у форматі знак|ціла частина.дробова частина, де знак 0 — додатне число, 1 — від'ємне. Дробова частина обчислюється з точністю до 3 розрядів.

**Алгоритм:**
1. Ціла частина: послідовне ділення на 2, залишки записуються у зворотному порядку.
2. Дробова частина: послідовне множення на 2, цілі частини результатів записуються по порядку.

```python
def task_7():
    FRAC_DIGITS = 3  # кількість розрядів дробової частини

    raw = input().strip()

    # --- контроль введених значень ---
    if raw == "":
        return "Помилка. Нічого не введено"
    try:
        number = float(raw.replace(",", "."))
    except ValueError:
        return "Помилка. Введене значення не є числом"
    if number != number or number in (float("inf"), float("-inf")):
        return "Помилка. Введене значення не є скінченним числом"

    sign = 1 if number < 0 else 0
    number = abs(number)

    integer_part = int(number)
    fractional_part = number - integer_part

    # --- ціла частина: ділення на 2 ---
    if integer_part == 0:
        int_bits = "0"
    else:
        bits = []
        n = integer_part
        while n > 0:
            bits.append(str(n % 2))
            n //= 2
        int_bits = "".join(reversed(bits))

    # --- дробова частина: множення на 2 ---
    result = int_bits
    is_fractional = "." in raw or "," in raw
    if is_fractional and fractional_part > 0:
        frac_bits = []
        f = fractional_part
        for _ in range(FRAC_DIGITS):
            f *= 2
            bit = int(f)
            frac_bits.append(str(bit))
            f -= bit
        result += "." + "".join(frac_bits)

    shown = raw.replace(",", ".")
    return f"'{shown} у двійковій системі буде {sign}|{result}'"
```
